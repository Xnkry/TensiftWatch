# Fill rate distribution — Lalla Takerkoust

Explore the available daily observations; missing fill rates are excluded.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/processed/tensift_cleaned.csv", parse_dates=["date"])
fill_rates = df.loc[df["dam"] == "Lalla Takerkoust", "fill_rate"].dropna()
print(fill_rates.describe())
```

```python
plt.figure(figsize=(9, 5))
plt.hist(fill_rates, bins=15, edgecolor="white")
plt.title("Daily Fill Rate Distribution - Lalla Takerkoust")
plt.xlabel("Fill Rate (%)")
plt.ylabel("Number of observations")
plt.tight_layout()
plt.show()
```
