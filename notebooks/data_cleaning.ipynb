```python

import pandas as pd
from pathlib import Path

file = "../data/raw/data_barrages_tensift_aout_2026.xlsx"

df = pd.read_excel(file, header=None)

df

```

```python
# for dam Lalla Takerkoust
dam = df.iloc[0, 3]
capacite = df.iloc[1, 3]

jours = df.iloc[3:, 0].reset_index(drop=True)
reserve = df.iloc[3:, 3].reset_index(drop=True)
taux = df.iloc[3:, 4].reset_index(drop=True)

clean_data = pd.DataFrame({
    "jours": jours,
    "dam": dam,
    "capacite": capacite,
    "reserve": reserve,
    "taux_remplissage": taux
})

print(clean_data)
```

```python
dams_data = []

# 3. Loop through the five dams
for i in range(1, 11, 2):

    # Extract information
    dam_name = df.iloc[0, i].strip()
    capacite = df.iloc[1, i]

    jours = df.iloc[3:, 0].reset_index(drop=True)
    reserve = df.iloc[3:, i].reset_index(drop=True)
    taux = df.iloc[3:, i + 1].reset_index(drop=True)

   
    clean_data = pd.DataFrame({
        "jour": pd.to_numeric(jours),
        "dam": dam_name,
        "capacity_mm3": capacite,
        "reserve_mm3": pd.to_numeric(reserve),
        "fill_rate": pd.to_numeric(taux)
    })
    clean_data["date"] = pd.to_datetime({
        "year": 2026,
        "month": 8,
        "day": clean_data["jour"]
    })
    clean_data = clean_data[
        ["date", "dam", "capacity_mm3",
         "reserve_mm3", "fill_rate"]
    ]
    dams_data.append(clean_data)

all_dams = pd.concat(dams_data, ignore_index=True)
print("Shape:", all_dams.shape)
print("\nFirst rows:")
print(all_dams.head())
print("\nMissing values:")
print(all_dams.isna().sum())
output = Path("../data/processed/tensift_august_2026.csv")
output.parent.mkdir(parents=True, exist_ok=True)

all_dams.to_csv(output, index=False)

print("\nDataset successfully exported!")
```
