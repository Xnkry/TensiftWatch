```python
import pandas as pd
import matplotlib.pyplot as plt
file = "../data/processed/tensift_cleaned.csv"
df = pd.read_csv(file, parse_dates=["date"])
lalla = df.loc[
    df["dam"] == "Lalla Takerkoust"
].sort_values("date")

plt.figure(figsize=(10, 5))

plt.plot(
    lalla["date"],
    lalla["reserve_mm3"],
    label="Lalla Takerkoust"
)

plt.title("Water Reserves Over Time")
plt.xlabel("Date")
plt.ylabel("Reserve (Mm³)")
plt.legend()
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

```

```python
averages = df.groupby("dam")["fill_rate"].mean()

plt.figure(figsize=(10, 5))

plt.bar(
    averages.index,
    averages.values
)

plt.title("Average Fill Rate by Dam")
plt.xlabel("Dam")
plt.ylabel("Average Fill Rate (%)")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```
